```python
from transformers import AutoModelForCausalLM, AutoTokenizer
```

```python
model = AutoModelForCausalLM.from_pretrained(
    "microsoft/Phi-3-mini-4k-instruct",
    device_map='cuda',
    torch_dtype='auto',
    trust_remote_code=True
)
tokenizer = AutoTokenizer.from_pretrained("microsoft/Phi-3-mini-4k-instruct")
```

```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    return_full_text=False,
    max_new_tokens=200,
    do_sample=False
)
```

```python
output1 = generator('Importance of artificial intelligence in education', use_cache=False)
print("\nOutput 1:\n", output1[0]["generated_text"])
```

```python
output2 = generator('Role of a tokenizer in a transformer-based language model', use_cache=False)
print("\nOutput 2:\n", output2[0]["generated_text"])
```

```python
output3 = generator('How habits shape our daily life', use_cache=False)
print("\nOutput 3:\n", output3[0]["generated_text"])

```
